```python
## What is the average age of patients in the dataset?
average_age  = df['age'].mean()
print('Average_Age:-',average_age)
```

Key Insight-- Average age of the patients are 54.42, indicating that most individuals in the dataset are middle-aged adults, suggesting heart disease assessment is more common in this age group.

```python
## What is the gender distribution of patients?
Gender_Distribution = df['sex'].value_counts()
print(Gender_Distribution)
```

Key Insight-- The dataset contains 206 male patients (68.2%) and 96 female patients (31.8%), indicating that male patients are represented more than twice as often as female patients.

```python
## What is the average resting blood pressure of patients?
average_resting_bp = df['trestbps'].mean()
print('Average_Resting_BP:-',average_resting_bp)
```

Key Insight-- The average resting blood pressure is 131.60 mmHg, suggesting that most patients have slightly higher than normal blood pressure levels.

```python
## How many patients have fasting blood sugar levels higher than 120 mg/dl?
fasting_bp = (df['fbs']== 1).sum()
print('Fasting_BP:-',fasting_bp)
```

Key Insight-- 45 patients have fasting blood sugar levels above 120 mg/dl, while most patients have normal fasting blood sugar levels.

```python
## What are the different types of chest pain recorded in the dataset?
cp_types = df['cp'].unique()
print('CP_Types:-',cp_types)
```

Key Insight-- There are 4 types of chest pain recorded in the dataset, showing that patients have different chest pain conditions.

```python
## What is the maximum heart rate achieved by patients?
max_heart_rate = df['thalach'].max()
print('Max_Heart_Rate:-',max_heart_rate)
```

Key Insight-- Maximum heart rate achieved by the patient is 202.

```python
## What percentage of patients experience exercise-induced angina?
exercise_induced_angina = df['exang'].mean()*100
print('Exercise_Induced_Angina:-',exercise_induced_angina)
```

Key Insight-- Around 32.78% of patients have exercise-induced angina, whereas most patients (67.22%) do not experience it.

```python
## What is the average cholesterol level in the dataset?
average_cholestrol = df['chol'].mean()
print('Average_Cholestrol:-',average_cholestrol)
```

Key Insight-- The average cholesterol level of patients is 246.5 mg/dl, indicating that the overall patient group has relatively high cholesterol levels, which may increase the risk of heart disease.

```python
## How many patients have a resting electrocardiographic result of 2?
ECG = df['restecg'].value_counts()[2] 
print('Number of patients with resting ECG result of 2:-',ECG)
```

```python
## What is the distribution of the number of major vessels colored by fluoroscopy?
# distribution of the number of major vessels (ca)
ca_distribution = df["ca"].value_counts().sort_index()
print(ca_distribution)
```

Key Insigth-- Most patients have 0 major vessels colored by fluoroscopy, while the number of patients decreases as the number of major vessels increases.

```python
## What is the correlation between age and cholesterol levels?
correlation = df[['age', 'chol']].corr()
print (correlation)
```

Key Insight-- There is a weak positive correlation (0.207) between age and cholesterol, showing that older patients tend to have slightly higher cholesterol levels.

```python
## What is the distribution of chest pain types across different age groups?
cp_age = df.groupby("age")["cp"].value_counts().unstack(fill_value=0)
print(cp_age)

plt.figure(figsize=(14,6))
cp_age.plot(kind="bar", stacked=True)
plt.title("Distribution of Chest Pain Types Across Different Age Groups")
plt.xlabel("Age")
plt.ylabel("Number of Patients")
plt.legend(title="Chest Pain Type")
plt.show()
```

Key Insight-- Different types of chest pain are observed across almost all age groups, showing that chest pain can occur at any age in the dataset.

```python
## How does maximum heart rate vary with exercise-induced angina?
heart_rate = df.groupby("exang")["thalach"].mean()
print(heart_rate)

plt.figure(figsize=(6,4))
sns.barplot(x="exang", y="thalach", data=df)
plt.title("Average Maximum Heart Rate by Exercise-Induced Angina")
plt.xlabel("Exercise-Induced Angina")
plt.ylabel("Average Maximum Heart Rate")
plt.xticks([0, 1], ["No", "Yes"])

plt.show()
```

Key Insight-- Patients without exercise-induced angina reached a higher average maximum heart rate (155.60 bpm) than those with exercise-induced angina (137.21 bpm).

```python
## Is there a significant difference in resting blood pressure between male and female patients?
Difference_Resting_BP = df.groupby('sex')['trestbps'].mean()
print(Difference_Resting_BP)
```

Key Insight-- Female patients have a slightly higher average resting blood pressure (133.08 mmHg) than male patients (130.91 mmHg). The difference is small, indicating that both genders have similar average resting blood pressure.

```python
## What is the relationship between fasting blood sugar levels and the presence of heart disease?
fbs_target = pd.crosstab(df["fbs"], df["target"])
print(fbs_target)
```

```python
plt.figure(figsize=(6,4))
pd.crosstab(df["fbs"], df["target"]).plot(kind="bar", stacked=True)
plt.title("Fasting Blood Sugar vs Heart Disease")
plt.xlabel("Fasting Blood Sugar (0 = ≤120, 1 = >120 mg/dl)")
plt.ylabel("Number of Patients")
plt.legend(["No Heart Disease", "Heart Disease"])

plt.show()
```

Key Insight-- Heart disease is present in patients with both normal and high fasting blood sugar levels, indicating a weak relationship between fasting blood sugar and heart disease.

```python
## How does the number of major vessels (ca) affect the target variable (heart disease presence)?
ca_target = pd.crosstab(df["ca"], df["target"])
print(ca_target)

plt.figure(figsize=(6,4))
pd.crosstab(df["ca"], df["target"]).plot(kind="bar", stacked=True)
plt.title("Major Vessels (ca) vs Heart Disease")
plt.xlabel("Number of Major Vessels (ca)")
plt.ylabel("Number of Patients")
plt.legend(["No Heart Disease", "Heart Disease"])

plt.show()
```

Key Insight-- Most patients with heart disease have 0 major vessels colored by fluoroscopy. Heart disease cases become less common as the number of major vessels increases.

```python
## What is the average oldpeak value for patients with different types of chest pain?
Average_Oldpeak =df.groupby('cp')['oldpeak'].mean()
print(Average_Oldpeak)
```

Key Insight-- Patients with chest pain type 3 have the highest average oldpeak (1.39), followed closely by chest pain type 0 (1.38). Chest pain type 1 has the lowest average oldpeak (0.32).

```python
## Analyze the distribution of thalassemia types (thal) among patients with heart disease.
thal_target = pd.crosstab(df["thal"], df["target"])
print(thal_target)
```

```python
plt.figure(figsize=(6,4))
pd.crosstab(df["thal"], df["target"]).plot(kind="bar", stacked=True)
plt.title("Thal Types vs Heart Disease")
plt.xlabel("Thal Type")
plt.ylabel("Number of Patients")
plt.legend(["No Heart Disease", "Heart Disease"])

plt.show()
```

Key Insight-- Thal type 2 has the highest number of patients with heart disease (129), while thal type 3 has the highest number of patients without heart disease (89). This suggests that thal type 2 is more commonly associated with heart disease in this dataset.

```python
## What are the most common combinations of risk factors in patients with heart disease?
risk_factors = (df[df["target"] == 1].groupby(["cp", "fbs", "exang", "thal"]).size().reset_index(name="Count").sort_values(by="Count", ascending=False))

print(risk_factors)
```

Key Insight-- 41 patients with heart disease have the combination of Chest Pain Type 2, normal fasting blood sugar, no exercise-induced angina, and Thalassemia Type 2, making it the most common risk factor combination.

```python
## Perform a pairwise comparison of clinical measurements for patients with and without heart disease.
heart_disease = df[df["target"] == 1].describe()
print(heart_disease)
```
